```python
import pandas as pd
data = {
    "device_id": ["00815", "00815", "0231", "00815", "0231",
                 "4020", "4020", "0231", "00815", "4020"],
    "date": ["2024-03-01", "2024-03-01", "2024-03-01",
             "2024-03-02", "2024-03-02", "2024-03-02",
             "2024-03-02", "2024-03-03", "2024-03-03",
             "2024-03-01"],
    "reading": ["42", "46", "39", "51", "N/A",
                "48", "", "44", "45", "40"]
}

df = pd.DataFrame(data)

df

df["reading_numeric"] = pd.to_numeric(
    df["reading"],
    errors="coerce"
)

summary = (
    df.groupby("date")["reading_numeric"]
    .agg(["count", "mean"])
    .reset_index()
)
summary.columns = [
    "date",
    "valid_readings_count",
    "average_reading"
]

summary
```

```python
import os

# Create output folder
os.makedirs("output", exist_ok=True)

# Create one CSV per device
for device_id, group in df.groupby("device_id"):
  group.to_csv(f"output/{device_id}.csv", index=False)

print("Device files created successfully!")

summary.to_csv("output/daily_summary.csv", index=False)
print("Daily summary file created!")
os.listdir("output")
pd.read_csv("output/00815.csv")
from google.colab import files
files.download("output/daily_summary.csv")
files.download("output/00815.csv")
files.download("output/0231.csv")
files.download("output/4020.csv")


```
